```python
import pandas as pd
import yfinance as yf
from datetime import datetime

#ticker tracking
tickers = ["NVDA", "BABA", "NTES", "GOOGL", "JD"]

#download data of the tickers from within a set period of time
data = yf.download(tickers, period="1y")

#Filter for closing prices
closingPrices = data['Close']
#round to 2 decimal places
closingPrices = closingPrices.round(2)
#format dates into ymd
closingPrices.index = closingPrices.index.strftime('%Y-%m-%d')
closingPrices.index.name = 'Date'

#create spreadsheet with the data
todayStr = datetime.now().strftime('%Y-%m-%d')
fileName = f"stock_closing_prices_{todayStr}.xlsx"
closingPrices.to_excel(fileName)
```
