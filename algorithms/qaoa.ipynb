-> A portfolio is built from a pool n of financial assets, with each asset labeled $ i \in {1,....,n} $

-> Every asset's return is a random variable with expected value and variance (This models the financial risk involved in the asset)

-> Every two assets (where they are not the same asset) have a covariance (modeling market correlation between assets)

-> Every asset i has $ w \in D = \{0,..,b_i\} $ with b_i defined as the budget for asset i (maxmum allotted money/weight for the asset)

-> return a vector in the covariance matrix, and the weight vector w are defined naturally from above 

for simplicity, assume all constants and variables are dimensionless so we can just find the optimal weight w where $$ \min_{w \in D} w^T \Sigma w \quad - \quad \mu^T w $$
subject to $\sum_{i}w_i <= B $

This case is called integer portfolio optimization, since the domans D_i are over positive integers.
