```python
import pandas as pd

stocks = [
    "BABA daily return",
    "GOOGL daily return",
    "JD daily return",
    "NTES daily return",
    "NVDA daily return"
]
```

```python
df = pd.read_csv('/content/Copy of stock_closing_prices_2026-11-02 - stock_closing_prices_2026-11-02 (1).csv')

df.head()
```

```python
# Take out summary row
df = df[df["Date"].astype(str).str.match(r"^\d{4}-\d{2}-\d{2}$")]

# Convert percentages into decimals
for stock in stocks:
    df[stock] = (
        pd.to_numeric(
            df[stock].astype(str).str.replace("%", ""),
            errors="coerce"
        ) / 100
    )

# Calculate sample variance
for stock in stocks:
    variance = df[stock].var()
    print(stock, "Variance:", variance)
```
